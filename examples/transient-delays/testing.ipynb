```python
import matplotlib.pyplot as plt
import numpy as np


def pdf(x, c, mu):
   return np.maximum(3/(4*c)*(1-np.pow((x-mu)/c, 2)), 0) 

c = 2
mu= 3

sampled = [2*c*np.sin(1/3*np.arcsin(2*np.random.uniform(0, 1)-1)) + mu for _ in range(100000)]
plt.hist(sampled, bins=500, density=True, alpha=0.5, color='g', label='Sampled')

x = np.linspace(-c-1+mu,c+1+mu, 1000)
plt.plot(x, pdf(x, c=c, mu=mu))
plt.show()
```
